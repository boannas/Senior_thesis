```python
import os
import itertools


# difficulty -> corresponding threshold
difficulty_thr = {
    'easy': 0,
    'normal': 1,
    'hard': 2,
}

g_values = [10, 15, 20]
i_values = [10, 15, 20]
a_values = ['0p3', '0p5', '0p7']


evolve = 'Evolved_results'
pl1 = os.path.join(os.getcwd(), evolve)

for diff, thr in difficulty_thr.items():

    test_path = os.path.join(pl1, diff)

    print(f'\n===== Checking {diff} (thr{thr}) =====')

    if not os.path.exists(test_path):
        print(f'Path not found: {test_path}')
        continue

    existing_folders = set(os.listdir(test_path))

    # generate expected folders
    expected_folders = set()

    for g, i, a in itertools.product(g_values, i_values, a_values):

        folder_name = (
            f'E2_thr{thr}_'
            f'g{g}_i{i}_a{a}_seed42'
        )

        expected_folders.add(folder_name)

    missing = sorted(expected_folders - existing_folders)
    valid_existing = sorted(expected_folders & existing_folders)

    print(f'Expected : {len(expected_folders)}')
    print(f'Found    : {len(existing_folders)}')
    print(f'Missing  : {len(missing)}')

    # missing expected folders
    if missing:
        print('\nMissing folders:')
        for m in missing:
            print(m)

    # check lineage_generations.csv
    print('\nChecking lineage_generations.csv ...')

    missing_csv = []

    for folder in valid_existing:

        folder_path = os.path.join(test_path, folder)

        csv_path = os.path.join(folder_path, 'lineage_generations.csv')

        if not os.path.isfile(csv_path):
            missing_csv.append(folder)

    if len(missing_csv) == 0:
        print('All folders contain lineage_generations.csv')
    else:
        print(f'Folders missing lineage_generations.csv ({len(missing_csv)}):')

        for folder in missing_csv:
            print(folder)
```
